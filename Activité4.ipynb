# Création de graphiques avec Matplotlib et Seaborn

Ce notebook explore différentes techniques de visualisation de données en utilisant Matplotlib et Seaborn. La visualisation des données est cruciale car elle permet de :
- Communiquer des idées claires
- Comprendre les tendances dans les données
- Identifier des patterns visuellement

```python
# Import des bibliothèques nécessaires
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

print(plt.style.available)
# Configuration du style
plt.style.use('ggplot')
sns.set_theme()
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Visualisations avec Matplotlib

### 1.1 Graphique en barres
Commençons par créer un graphique en barres simple pour comparer les ventes par région.

```python
# Données pour le graphique en barres
regions = ["Europe", "Asie", "Amérique"]
ventes = [3000, 5000, 4000]

plt.figure()
plt.bar(regions, ventes, color="skyblue")
plt.title("Ventes par région", fontsize=14, fontweight="bold")
plt.xlabel("Région", fontsize=12)
plt.ylabel("Ventes", fontsize=12)
plt.show()
```

### 1.2 Graphique linéaire
Visualisons maintenant la tendance des ventes mensuelles.

```python
# Données pour le graphique linéaire
mois = ["Jan", "Fév", "Mar", "Avr", "Mai"]
ventes_mensuelles = [1000, 1200, 1100, 1500, 1700]

plt.figure()
plt.plot(mois, ventes_mensuelles, marker="o", color="green", linestyle="--")
plt.title("Tendance des ventes mensuelles", fontsize=14, fontweight="bold")
plt.xlabel("Mois", fontsize=12)
plt.ylabel("Ventes", fontsize=12)
plt.grid(True)
plt.show()
```

### 1.3 Graphique circulaire
Représentons la répartition des ventes sous forme de camembert.

```python
plt.figure(figsize=(8, 8))
plt.pie(ventes, labels=regions, autopct="%1.1f%%", startangle=90,
        colors=["gold", "lightcoral", "lightskyblue"])
plt.title("Répartition des ventes par région", fontsize=14, fontweight="bold")
plt.axis("equal")
plt.show()
```

## 2. Visualisations avec Seaborn

### 2.1 Graphique en barres groupées
Création d'un graphique en barres plus élaboré avec Seaborn.

```python
# Données pour Seaborn
data = pd.DataFrame({
    "Région": ["Europe", "Asie", "Europe", "Asie", "Amérique", "Amérique"],
    "Ventes": [3000, 5000, 4000, 4500, 7000, 6500],
    "Trimestre": ["T1", "T1", "T2", "T2", "T1", "T2"]
})

plt.figure()
sns.barplot(x="Trimestre", y="Ventes", hue="Région", data=data, ci=None, palette="muted")
plt.title("Ventes trimestrielles par région", fontsize=14, fontweight="bold")
plt.show()
```

### 2.2 Nuage de points
Analysons la relation entre le budget publicitaire et les ventes.

```python
# Données pour le nuage de points
budget_ventes = pd.DataFrame({
    "Budget_Publicité": [200, 300, 400, 500, 600],
    "Ventes": [1000, 1500, 2000, 2500, 3000],
    "Catégorie": ["Low", "Medium", "Medium", "High", "High"]
})

plt.figure()
sns.scatterplot(x="Budget_Publicité", y="Ventes", size="Ventes",
                hue="Catégorie", data=budget_ventes, sizes=(50, 300), palette="cool")
plt.title("Relation entre budget publicitaire et ventes", fontsize=14, fontweight="bold")
plt.show()
```

### 2.3 Visualisations statistiques
Création d'un boxplot et d'une heatmap pour analyser la distribution et les corrélations.

```python
# Boxplot
plt.figure()
sns.boxplot(x="Région", y="Ventes", data=data, palette="Set2")
plt.title("Distribution des ventes par région", fontsize=14, fontweight="bold")
plt.show()
numerical_data = data.select_dtypes(include=["float64", "int64"])  # Garder uniquement les colonnes numériques

# Calcul de la corrélation
corr_matrix = numerical_data.corr()
# Heatmap
plt.figure(figsize=(8, 6))
corr_matrix = budget_ventes.corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matrice de corrélation des métriques", fontsize=14, fontweight="bold")
plt.show()
```

## 3. Exercices pratiques

### Exercice 1: Analyse des ventes trimestrielles

```python
# Données pour les exercices
data_exercice = pd.DataFrame({
    "Région": ["Europe", "Asie", "Amérique", "Europe", "Asie", "Amérique"],
    "Trimestre": ["T1", "T1", "T1", "T2", "T2", "T2"],
    "Ventes": [3000, 5000, 7000, 3500, 4800, 7200],
    "Croissance": [5, 7, 8, 6, 7, 10]
})

plt.figure()
sns.barplot(x="Trimestre", y="Ventes", hue="Région",
            data=data_exercice, palette="viridis")
plt.title("Ventes trimestrielles par région", fontsize=14, fontweight="bold")
plt.show()
```

### Exercice 2: Analyse des corrélations

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Exemple de nettoyage des données
numerical_data = data_exercice.select_dtypes(include=["float64", "int64"])  # Garder uniquement les colonnes numériques

# Calcul de la corrélation
corr_exercice = numerical_data.corr()

# Création de la heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr_exercice, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Corrélation entre variables", fontsize=14, fontweight="bold")
plt.show()
```

### Exercice 3: Analyse de la distribution

```python
plt.figure()
sns.boxplot(x="Région", y="Ventes", data=data_exercice, palette="cool")
plt.title("Distribution des ventes", fontsize=14, fontweight="bold")
plt.show()
```

## Conclusion
