# Analyse des Ventes Régionales

Ce rapport explore les tendances des ventes trimestrielles par région afin d'identifier les opportunités de croissance.

````python
# Import des bibliothèques nécessaires
import pandas as pd
import matplotlib.pyplot as plt

# Création du dataset fictif
data = {
    'Region': ['Europe', 'Asie', 'Amérique', 'Afrique'],
    'Ventes': [120000, 80000, 100000, 50000]
}

# Création du DataFrame
df = pd.DataFrame(data)
````

````python
# Création du graphique à barres
plt.figure(figsize=(10, 6))
df.plot(kind='bar', x='Region', y='Ventes', title='Ventes par Région')
plt.xlabel('Région')
plt.ylabel('Ventes (en $)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

## Analyse des Ventes

Le graphique montre que l'Europe est la région avec les ventes les plus élevées, tandis que l'Afrique a les ventes les plus faibles. Cette visualisation nous permet d'identifier clairement les marchés les plus performants et ceux qui nécessitent peut-être plus d'attention ou d'investissement.
