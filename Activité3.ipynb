# Transformation des données et analyse statistique

Ce notebook présente différentes techniques de transformation de données et d'analyse statistique avec Pandas.

## Préparation des données

Commençons par importer Pandas et créer un DataFrame avec des données fictives.

```python
import pandas as pd

# Création du DataFrame avec des données fictives
data = {
    "Region": ["Europe", "Asie", "Amérique", "Europe", "Afrique", "Asie"],
    "Mois": ["Janvier", "Janvier", "Février", "Février", "Mars", "Mars"],
    "Ventes": [1200, 800, 1500, 1300, 500, 900]
}

df = pd.DataFrame(data)
print("DataFrame initial :")
print(df)
```

## 1. Création d'une colonne basée sur une condition

Nous allons créer une nouvelle colonne 'Performance' qui catégorise les ventes selon un seuil de 1000.

```python
# Création de la colonne Performance
df["Performance"] = df["Ventes"].apply(lambda x: "Bonne" if x > 1000 else "Moyenne")

print("DataFrame avec la colonne 'Performance':")
print(df)
```

## 2. Transformation mathématique

Normalisons les ventes en utilisant la moyenne et l'écart-type (z-score).

```python
# Normalisation des ventes
df["Ventes_normalisees"] = (df["Ventes"] - df["Ventes"].mean()) / df["Ventes"].std()

print("DataFrame avec les ventes normalisées :")
print(df)
```

## 3. Regroupement simple

Calculons la somme des ventes par région.

```python
# Calcul des ventes totales par région
ventes_par_region = df.groupby("Region")["Ventes"].sum()

print("Somme des ventes par région :")
print(ventes_par_region)
```

## 4. Agrégations multiples

Calculons plusieurs statistiques par région : moyenne, somme et nombre d'entrées.

```python
# Calcul de statistiques multiples par région
stats_region = df.groupby("Region").agg({
    "Ventes": ["mean", "sum", "count"]
})

# Renommage des colonnes pour plus de clarté
stats_region.columns = ["Moyenne", "Total", "Nombre"]

print("Statistiques par région :")
print(stats_region)
```

## 5. Regroupement avancé (multi-niveaux)

Analysons les ventes moyennes par région et par mois.

```python
# Calcul des moyennes par région et mois
stats_avancees = df.groupby(["Region", "Mois"])["Ventes"].mean()

print("Moyenne des ventes par région et mois :")
print(stats_avancees)

# Affichage sous forme de tableau croisé pour une meilleure lisibilité
print("\nTableau croisé des ventes moyennes :")
print(pd.pivot_table(df, values="Ventes", index="Region", columns="Mois", aggfunc="mean"))
```
