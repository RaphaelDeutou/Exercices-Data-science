# Analyse des poissons 

## Importation des données


```python
import pandas as pd
df = pd.read_csv("fish_catches.csv")
print(df.head())
```

## Maintenant nous devons nettoyer les données

### Supprimons les lignes avec des valeurs manquantes

```python
df = df.dropna(axis=1, how='all')

print("\nAprès suppression des colonnes vides :")
print(df)
```

### On supprime les lignes NaN

```python
df["Total_poisson"] = (df["2006"] + df["2007"]+ df["2008"]+ df["2009"]+ df["2010"]+ df["2011"]+ df["2012"]+ df["2013"]+ df["2014"])
print(df)
```

### Je montres maintenant la performance des pêches

```python
df["Performance"] = df["Total_poisson"].apply(lambda x: "Bonne" if x >
500 else "Moyenne")
print(df)
```

### Le poisson le plus peché

### je commence par faire une selection 

```python
df_selection = df[["Species", "Total_poisson"]]

print(df_selection)
```

```python
df_sorted= df.sort_values(by="Total_poisson", ascending=False)

print(df_sorted)
```

```python
# Grouper par 'Species' et sommer les données par année
grouped = df.groupby("Species")[["2014", "2013", "2012", "2011", "2010", "2009", "2008", "2007", "2006"]].sum()



print("\nTotaux des ventes par espèce et par année :")
print(grouped)
```
