# Chargement et manipulation des données avec Pandas

Ce notebook présente les différentes méthodes pour charger et inspecter des données avec Pandas à partir de différentes sources : CSV, Excel, JSON et API.

## 1. Import des bibliothèques

Commençons par importer les bibliothèques nécessaires pour notre analyse.

```python
import pandas as pd
import requests

print("Versions des bibliothèques :")
print(f"pandas version: {pd.__version__}")
```

## 2. Chargement d'un fichier CSV

Nous allons charger un fichier CSV contenant des données de ventes. Le fichier contient les colonnes : Date, Produit, Region, Ventes.

```python
# Chargement du fichier CSV
df_csv = pd.read_csv("ventes.csv")

print("Aperçu des 5 premières lignes du CSV :")
print(df_csv.head())
```

## 3. Chargement d'un fichier Excel

Maintenant, nous allons charger des données depuis un fichier Excel, en spécifiant une feuille particulière.

```python
# Chargement du fichier Excel
df_excel = pd.read_excel("ventes.xlsx", sheet_name="Janvier")

print("Aperçu des 5 premières lignes du fichier Excel :")
print(df_excel.head())
```

## 4. Chargement d'un fichier JSON

Les données peuvent également être stockées au format JSON. Voici comment les charger.

```python
# Chargement du fichier JSON
df_json = pd.read_json("ventes.json")

print("Aperçu des 5 premières lignes du fichier JSON :")
print(df_json.head())
```

## 5. Importation via API

Il est également possible de récupérer des données directement depuis une API.

```python
# Récupération des données via API
url = "https://api.example.com/ventes"
response = requests.get(url)

if response.status_code == 200:
    df_api = pd.DataFrame(response.json())
    print("Aperçu des 5 premières lignes des données API :")
    print(df_api.head())
else:
    print("Erreur lors de la récupération des données via l'API.")
```

## 6. Inspection initiale des données

Une fois les données chargées, il est important de les inspecter pour comprendre leur structure et leur qualité.

```python
# Informations sur le dataset CSV
print("Informations sur le dataset CSV :")
print(df_csv.info())

print("\nStatistiques descriptives du dataset CSV :")
print(df_csv.describe())
```

## 7. Identification des problèmes

Nous alons maintenant identifier les problèmes potentiels dans les données comme les valeurs manquantes ou les doublons.

```python
# Vérification des valeurs manquantes et des doublons
print("Valeurs manquantes dans le dataset CSV :")
print(df_csv.isnull().sum())

print("\nDoublons dans le dataset CSV :")
print(df_csv.duplicated().sum())
```
